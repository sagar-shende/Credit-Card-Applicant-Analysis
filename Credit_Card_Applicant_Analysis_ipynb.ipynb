# Insight

# Insight
Only the Occupation Type column had missing values, while all other columns were complete. The missing values were replaced with "Other" so that no customer records were removed during the analysis.

```python
# Feature Engineering
```

```python
df["Age"] = (df["DAYS_BIRTH"].abs()/365).astype(int)
```

```python
df.head(2)
```

```python
df["Age_Group"] = pd.cut(df["Age"], bins = [ 19, 29, 44, 59, 100],
labels =[ "Young Adult", "Adult", "Middle Age", "Senior"])
```

```python
df.head(2)
```

```python
df["Employment_Years"] = (df["DAYS_EMPLOYED"].replace(365243, 0))
df["Employment_Years"] = (df["Employment_Years"].abs()/365).astype(int)
```

```python
df.head(10)
```

```python
df["IS_FLAG_MOBIL"] = np.where(df["FLAG_MOBIL"] == 1, "Yes", "No")
df["IS_FLAG_WORK_PHONE"] = np.where(df["FLAG_WORK_PHONE"] == 1, "Yes", "No")
df["IS_FLAG_PHONE"] = np.where(df["FLAG_PHONE"] == 1, "Yes", "No")
df["Is_FLAG_OWN_CAR"] = np.where(df["FLAG_OWN_CAR"] == "Y", "1", "0")
df["Is_FLAG_OWN_REALTY"] = np.where(df["FLAG_OWN_REALTY"] == "Y", "1", "0")
```

```python
df.head(2)
```

# Insight
New columns such as Age, Age Group, and Employment Years were created to make customer analysis easier. Binary values were also converted into readable Yes/No categories.

```python
#Q10- What is the relationship between education level and occupation?
education_occupation = pd.crosstab(df["NAME_EDUCATION_TYPE"], df["OCCUPATION_TYPE"])
```

```python
education_occupation
```

```python
plt.figure(figsize = (12,6))
sns.heatmap(education_occupation , annot = True,  fmt=".0f",
    cmap="viridis",
    linewidths=0.5,
    annot_kws={"size":8}
)

plt.title("Education vs Occupation Heatmap", fontsize=14)
plt.xlabel("Education Level")
plt.ylabel("Occupation")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

# Insight
Customers with Secondary / Secondary Special education form the largest group across most occupations, particularly in the "Other" occupation category. Customers with Higher Education are also well represented, especially in Core Staff, Managers, Medicine Staff, and Sales Staff roles. In contrast, Academic Degree and Lower Secondary education levels have relatively few customers across all occupations. This indicates that the customer base is primarily composed of applicants with secondary and higher education backgrounds, which can help financial institutions design targeted products and customer engagement strategies.

```python
#2- How does family size vary across family status and housing type?
family_status_housing = df.groupby(["NAME_FAMILY_STATUS", "NAME_HOUSING_TYPE"]) ["CNT_FAM_MEMBERS"].mean().sort_values(ascending = False).unstack()
```

```python
family_status_housing
```

```python
plt.figure(figsize = (10,6))
sns.heatmap(family_status_housing, annot = True, cmap = "coolwarm")
plt.title("Family Status vs Housing Type Heatmap")
plt.show()
```

# Insight
Married customers generally have larger family sizes across all housing types, while Single and Widowed customers have smaller households.

```python
#3- Which income type has the highest work phone availability across occupations?
work_phone = df.groupby(["OCCUPATION_TYPE", "NAME_INCOME_TYPE"]) ["FLAG_WORK_PHONE"].mean().sort_values(ascending = False).unstack().fillna(0)
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    work_phone,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Work Phone Availability Across Occupations and Income Types")
plt.xlabel("Income Type")
plt.ylabel("Occupation")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

# Insight
State Servants show the highest work phone availability, indicating stronger formal employment compared with other occupation groups.
